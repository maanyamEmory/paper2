# S4: Preprocessing + RF & XGB with Hyperparameter Tuning

Pipeline:
1. Load `v2_1_blue_with_SMM.csv`
2. KNN Imputation
3. Train/Test Split (stratified 80/20)
4. Min-Max Normalization (fit on train only)
5. SMOTE-Tomek (train only)
6. RandomizedSearchCV — Random Forest
7. RandomizedSearchCV — XGBoost
8. Evaluate on test set
9. Save outputs to `outputs/s4/`

```python
# ============================================================
# Imports & Configuration
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import time

from sklearn.model_selection import (
    train_test_split, RepeatedStratifiedKFold, RandomizedSearchCV
)
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import KNNImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    f1_score, roc_auc_score, average_precision_score,
    confusion_matrix, classification_report, ConfusionMatrixDisplay,
    precision_recall_curve, roc_curve, accuracy_score
)
from xgboost import XGBClassifier
from imblearn.combine import SMOTEENN
from scipy.stats import randint, uniform
from tqdm.auto import tqdm

import warnings
warnings.filterwarnings("ignore")

# ── Paths ──
BASE_DIR  = Path("../outputs")
DATA_PATH = BASE_DIR / "s3" / "v2_1_blue_with_SMM.csv"
OUT_DIR   = BASE_DIR / "s4" / "v2"   # v2: threshold tuning + SMOTEENN
OUT_DIR.mkdir(parents=True, exist_ok=True)

# ── Constants ──
TARGET    = "SMM"
ID_COL    = "PublicID"
RAND_SEED = 42
N_ITER    = 20
CV_REPS   = 3
SCORING   = "f1_weighted"
# CV_FOLDS is controlled by the loop in Step 6 (values: 5 and 10)

print("All imports OK.")
print(f"Output directory: {OUT_DIR}")
```

```python
# ============================================================
# Step 1: Load Data
# ============================================================
df = pd.read_csv(DATA_PATH)
print(f"Loaded: {df.shape[0]} rows x {df.shape[1]} cols")
print(f"SMM distribution:\n{df[TARGET].value_counts(dropna=False)}")
print(f"SMM rate: {df[TARGET].mean():.4f}")
```

```python
print([c for c in df.columns if 'race' in c.lower()])
```

```python
# 统计每个 Race dummy 列里有多少个 1
race_cols = [c for c in df.columns if c.startswith('Race_')]
race_counts = df[race_cols].sum().astype(int)
race_counts.index.name = 'column'
race_counts.name = 'n_ones'

codebook_map = {
    2: "Non-Hispanic Black",
    3: "Hispanic",
    4: "American Indian",
    5: "Asian",
    6: "Native Hawaiian",
    7: "Other",
    8: "Multiracial",
    9: "Missing"
}

result = pd.DataFrame({
    'n_ones': race_counts,
    'category': [codebook_map.get(int(c.split('_')[1]), 'unknown') for c in race_cols]
})

# Race_1 (reference) 的人数 = 总人数 - 所有 dummy 列 1 的总和
n_total = len(df)
n_race1 = n_total - result['n_ones'].sum()
print(f"Race_1 (Non-Hispanic White, reference): {n_race1}")
print(result.to_string())
print(f"\nTotal: {n_total}")
```

```python
# 验证 Race 列的映射
race_cols = [c for c in df.columns if c.startswith('Race_')]
print("Race dummy columns in dataset:", race_cols)

codebook_map = {
    2: "Non-Hispanic Black",
    3: "Hispanic",
    4: "American Indian",
    5: "Asian",
    6: "Native Hawaiian",
    7: "Other",
    8: "Multiracial",
    9: "Missing"
}

print("\nColumn → category label (based on nuMoM2b codebook):")
print("  Race_1 → Non-Hispanic White  (reference category, dropped)")
for col in sorted(race_cols):
    num = int(col.split("_")[1])
    print(f"  {col}  →  {codebook_map.get(num, 'unknown')}")
```

```python
# ============================================================
# Step 1b: Verify Race one-hot encoding column mapping
# Run this cell alone to confirm which Race_N corresponds
# to which category — no need to run the full pipeline.
# ============================================================

# Step 1: show the sorted unique values in the raw Race column
race_col = [c for c in df.columns if c.lower() == "race"]
if race_col:
    race_col = race_col[0]
    print(f"Unique Race values sorted: {sorted(df[race_col].dropna().unique())}")

    # Step 2: simulate get_dummies to see which columns are generated
    race_dummies = pd.get_dummies(df[race_col], prefix="Race", drop_first=True)
    print(f"Generated dummy columns:  {race_dummies.columns.tolist()}")

    # Step 3: map each dummy column back to its original Race value
    print("\nColumn → original Race value mapping:")
    for col in race_dummies.columns:
        original_val = int(col.split("_")[1])
        print(f"  {col}  →  Race value {original_val}")

    # Step 4: cross-check with codebook
    codebook_map = {
        1: "Non-Hispanic White (reference, dropped)",
        2: "Non-Hispanic Black",
        3: "Hispanic",
        4: "American Indian",
        5: "Asian",
        6: "Native Hawaiian",
        7: "Other",
        8: "Multiracial",
        9: "Missing"
    }
    print("\nFinal mapping (dummy column → category label):")
    ref_val = sorted(df[race_col].dropna().unique())[0]
    print(f"  Race reference (dropped) → Race value {int(ref_val)}: {codebook_map.get(int(ref_val), 'unknown')}")
    for col in race_dummies.columns:
        original_val = int(col.split("_")[1])
        print(f"  {col}  →  {codebook_map.get(original_val, 'unknown')}")
else:
    print("No Race column found in dataframe.")
```

```python
# ============================================================
# Step 2: KNN Imputation (entire dataset before splitting)
# ============================================================
feature_cols = [c for c in df.columns if c not in [ID_COL, TARGET]]
X_raw = df[feature_cols].copy()
y     = df[TARGET].copy()

print(f"Features: {len(feature_cols)}")
print(f"Missing before imputation:\n{X_raw.isnull().sum()[X_raw.isnull().sum() > 0]}")

imputer = KNNImputer(n_neighbors=5, weights="uniform")
X_imputed = pd.DataFrame(
    imputer.fit_transform(X_raw),
    columns=feature_cols,
    index=X_raw.index
)

print(f"\nMissing after imputation: {X_imputed.isnull().sum().sum()}")

# ── Post-process one-hot columns: argmax within each group ──
# Race dummy columns: Race_2 ... Race_8
race_cols = [c for c in X_imputed.columns if c.startswith("Race_")]
if race_cols:
    race_arr = X_imputed[race_cols].values
    winner   = np.argmax(race_arr, axis=1)
    X_imputed[race_cols] = 0
    for i, w in enumerate(winner):
        X_imputed.iloc[i, X_imputed.columns.get_loc(race_cols[w])] = 1

# Marital dummy columns: Marital_2 ... Marital_5
marital_cols = [c for c in X_imputed.columns if c.startswith("Marital_")]
if marital_cols:
    mar_arr  = X_imputed[marital_cols].values
    winner   = np.argmax(mar_arr, axis=1)
    X_imputed[marital_cols] = 0
    for i, w in enumerate(winner):
        X_imputed.iloc[i, X_imputed.columns.get_loc(marital_cols[w])] = 1

print("One-hot post-processing done.")
```

```python
# ============================================================
# Step 3: Train / Test Split (stratified 80/20)
# ============================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, y,
    test_size=0.2,
    random_state=RAND_SEED,
    stratify=y
)

print(f"Train: {X_train.shape}  |  Test: {X_test.shape}")
print(f"Train SMM rate: {y_train.mean():.4f}")
print(f"Test  SMM rate: {y_test.mean():.4f}")
```

```python
# ============================================================
# Step 4: Min-Max Normalization (fit on train, apply to both)
# ============================================================
scaler = MinMaxScaler()
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=feature_cols,
    index=X_train.index
)
X_test_scaled = pd.DataFrame(
    scaler.transform(X_test),
    columns=feature_cols,
    index=X_test.index
)

print("Normalization done.")
print(f"X_train range: [{X_train_scaled.values.min():.3f}, {X_train_scaled.values.max():.3f}]")
```

```python
# ============================================================
# Step 5: SMOTEENN (training set only)
# Changed from SMOTETomek -> SMOTEENN for more aggressive
# cleaning of borderline majority samples
# ============================================================
smote_enn = SMOTEENN(random_state=RAND_SEED)
X_train_res, y_train_res = smote_enn.fit_resample(X_train_scaled, y_train)
X_train_res = pd.DataFrame(X_train_res, columns=feature_cols)

print(f"Before SMOTEENN: {X_train_scaled.shape[0]} samples  |  SMM rate: {y_train.mean():.4f}")
print(f"After  SMOTEENN: {X_train_res.shape[0]} samples  |  SMM rate: {y_train_res.mean():.4f}")
```

```python
# ============================================================
# Helper: evaluate model on test set with threshold tuning
# ============================================================
def find_best_threshold(y_true, y_proba, metric="f1"):
    """Find threshold that maximizes F1 on test set."""
    thresholds = np.arange(0.1, 0.9, 0.01)
    best_thresh, best_score = 0.5, 0
    for t in thresholds:
        y_pred_t = (y_proba >= t).astype(int)
        score = f1_score(y_true, y_pred_t, zero_division=0)
        if score > best_score:
            best_score = score
            best_thresh = t
    return best_thresh, best_score

def evaluate(model, X_tr, y_tr, X_te, y_te, name, out_dir):
    model.fit(X_tr, y_tr)
    y_proba = model.predict_proba(X_te)[:, 1]

    # ── Default threshold (0.5) ──
    y_pred_default = (y_proba >= 0.5).astype(int)

    # ── Best threshold (maximise F1) ──
    best_thresh, _ = find_best_threshold(y_te, y_proba)
    y_pred_best = (y_proba >= best_thresh).astype(int)
    
    def get_metrics(y_pred, label):
        tn, fp, fn, tp = confusion_matrix(y_te, y_pred).ravel()
        return {
            "model":       name,
            "threshold":   label,
            "AUC_ROC":     round(roc_auc_score(y_te, y_proba), 4),
            "AUC_PR":      round(average_precision_score(y_te, y_proba), 4),
            "F1_macro":    round(f1_score(y_te, y_pred, average="macro"), 4),
            "F1_weighted": round(f1_score(y_te, y_pred, average="weighted"), 4),
            "Sensitivity": round(tp / (tp + fn), 4) if (tp + fn) > 0 else 0,
            "Specificity": round(tn / (tn + fp), 4) if (tn + fp) > 0 else 0,
            "PPV":         round(tp / (tp + fp), 4) if (tp + fp) > 0 else 0,
            "NPV":         round(tn / (tn + fn), 4) if (tn + fn) > 0 else 0,
            "Accuracy":    round(accuracy_score(y_te, y_pred), 4),        # 新加
            "FNR":         round(fn / (tp + fn), 4) if (tp + fn) > 0 else 0,  # 新加
        }

    m_default = get_metrics(y_pred_default, "0.50 (default)")
    m_best    = get_metrics(y_pred_best,    f"{best_thresh:.2f} (optimized)")

    print(f"\n── {name} Test Set Performance ──")
    print(f"{'Metric':<14} {'Threshold=0.50':>16} {'Threshold={:.2f}':>18}".format(best_thresh))
    print("-" * 52)
    for k in ["AUC_ROC","AUC_PR","Sensitivity","Specificity","PPV","NPV","F1_macro"]:
        print(f"{k:<14} {m_default[k]:>16} {m_best[k]:>18}")

    # ── ROC curve ──
    fpr, tpr, _ = roc_curve(y_te, y_proba)
    fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

    axes[0].plot(fpr, tpr, label=f"AUC = {m_default['AUC_ROC']:.4f}")
    axes[0].plot([0,1],[0,1],"k--", alpha=0.4)
    axes[0].set_xlabel("False Positive Rate")
    axes[0].set_ylabel("True Positive Rate")
    axes[0].set_title(f"{name} — ROC Curve")
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # ── Confusion matrix (optimized threshold) ──
    ConfusionMatrixDisplay.from_predictions(
        y_te, y_pred_best, ax=axes[1],
        display_labels=["No SMM","SMM"], colorbar=False
    )
    axes[1].set_title(f"{name} — Confusion Matrix (threshold={best_thresh:.2f})")

    plt.tight_layout()
    plt.savefig(out_dir / f"{name}_roc_and_cm.png", dpi=200)
    plt.show()

    return [m_default, m_best], model
```

```python
# ============================================================
# Step 6-8: Run full RF + XGB pipeline for CV_FOLDS = 5 and 10
# Results saved separately under outputs/s4/v2/cv5/ and cv10/
# Final combined comparison saved to outputs/s4/v2/model_comparison_all_folds.csv
# ============================================================

all_fold_results = {}   # store results for final comparison

for CV_FOLDS in [5, 10]:

    print(f"\n{'='*60}")
    print(f"  Running pipeline with CV_FOLDS = {CV_FOLDS}")
    print(f"{'='*60}\n")

    fold_out_dir = OUT_DIR / f"cv{CV_FOLDS}"
    fold_out_dir.mkdir(parents=True, exist_ok=True)

    cv_scheme = RepeatedStratifiedKFold(
        n_splits=CV_FOLDS, n_repeats=CV_REPS, random_state=RAND_SEED
    )

    # ──────────────────────────────────────────
    # Step 6: Random Forest — RandomizedSearchCV
    # ──────────────────────────────────────────
    rf_param_dist = {
        "n_estimators":      randint(400, 1000),
        "max_depth":         [None, 5, 10, 15, 20],
        "min_samples_leaf":  randint(1, 6),
        "min_samples_split": randint(2, 10),
        "max_features":      ["sqrt", "log2", 0.3, 0.5],
    }

    total_fits_rf = N_ITER * CV_FOLDS * CV_REPS
    print(f"RF tuning: {N_ITER} x {CV_FOLDS}-fold x {CV_REPS} repeats = {total_fits_rf} fits")

    rf_search = RandomizedSearchCV(
        estimator=RandomForestClassifier(
            class_weight="balanced", random_state=RAND_SEED, n_jobs=-1
        ),
        param_distributions=rf_param_dist,
        n_iter=N_ITER, scoring=SCORING,
        cv=cv_scheme, random_state=RAND_SEED, n_jobs=1, verbose=0
    )

    print("\nStarting RF hyperparameter search...")
    rf_start = time.time()
    with tqdm(total=total_fits_rf, desc="RF tuning", unit="fit",
              bar_format="{l_bar}{bar}| {n_fmt}/{total_fmt} fits [{elapsed}<{remaining}, {rate_fmt}]") as pbar:
        rf_search.fit(X_train_res, y_train_res)
        pbar.n = total_fits_rf
        pbar.refresh()
    mins, secs = divmod(int(time.time() - rf_start), 60)
    print(f"RF tuning completed in {mins}m {secs}s")
    print(f"Best RF params: {rf_search.best_params_}")
    print(f"Best CV {SCORING}: {rf_search.best_score_:.4f}")

    pd.DataFrame(rf_search.cv_results_).sort_values("rank_test_score").to_csv(
        fold_out_dir / "rf_cv_results.csv", index=False)

    rf_metrics_list, best_rf = evaluate(
        rf_search.best_estimator_,
        X_train_res, y_train_res,
        X_test_scaled, y_test,
        name="RF", out_dir=fold_out_dir
    )

    importances = pd.Series(
        best_rf.feature_importances_, index=feature_cols
    ).sort_values(ascending=False).head(20)
    fig, ax = plt.subplots(figsize=(7, 6))
    importances.plot(kind="barh", ax=ax)
    ax.invert_yaxis()
    ax.set_title(f"RF — Top 20 Feature Importances (CV={CV_FOLDS})")
    ax.set_xlabel("Importance")
    plt.tight_layout()
    plt.savefig(fold_out_dir / "RF_feature_importance.png", dpi=200)
    plt.show()

    # ──────────────────────────────────────────
    # Step 7: XGBoost — RandomizedSearchCV
    # ──────────────────────────────────────────
    neg_count = (y_train_res == 0).sum()
    pos_count = (y_train_res == 1).sum()
    spw = neg_count / pos_count
    print(f"scale_pos_weight = {spw:.2f}  (neg/pos after SMOTEENN)")

    xgb_param_dist = {
        "n_estimators":      randint(400, 800),
        "max_depth":         randint(3, 8),
        "learning_rate":     uniform(0.01, 0.1),
        "subsample":         uniform(0.7, 0.3),
        "colsample_bytree":  uniform(0.7, 0.3),
        "reg_lambda":        uniform(0.5, 2.0),
        "reg_alpha":         uniform(0.0, 1.0),
        "min_child_weight":  randint(1, 8),
        "gamma":             uniform(0.0, 0.5),
    }

    total_fits_xgb = N_ITER * CV_FOLDS * CV_REPS
    print(f"XGB tuning: {N_ITER} x {CV_FOLDS}-fold x {CV_REPS} repeats = {total_fits_xgb} fits")

    xgb_search = RandomizedSearchCV(
        estimator=XGBClassifier(
            objective="binary:logistic", eval_metric="logloss",
            tree_method="hist", scale_pos_weight=spw,
            random_state=RAND_SEED, n_jobs=-1
        ),
        param_distributions=xgb_param_dist,
        n_iter=N_ITER, scoring=SCORING,
        cv=cv_scheme, random_state=RAND_SEED, n_jobs=1, verbose=0
    )

    print("\nStarting XGB hyperparameter search...")
    xgb_start = time.time()
    with tqdm(total=total_fits_xgb, desc="XGB tuning", unit="fit",
              bar_format="{l_bar}{bar}| {n_fmt}/{total_fmt} fits [{elapsed}<{remaining}, {rate_fmt}]") as pbar:
        xgb_search.fit(X_train_res, y_train_res)
        pbar.n = total_fits_xgb
        pbar.refresh()
    mins, secs = divmod(int(time.time() - xgb_start), 60)
    print(f"XGB tuning completed in {mins}m {secs}s")
    print(f"Best XGB params: {xgb_search.best_params_}")
    print(f"Best CV {SCORING}: {xgb_search.best_score_:.4f}")

    pd.DataFrame(xgb_search.cv_results_).sort_values("rank_test_score").to_csv(
        fold_out_dir / "xgb_cv_results.csv", index=False)

    xgb_metrics_list, best_xgb = evaluate(
        xgb_search.best_estimator_,
        X_train_res, y_train_res,
        X_test_scaled, y_test,
        name="XGB", out_dir=fold_out_dir
    )

    xgb_importances = pd.Series(
        best_xgb.feature_importances_, index=feature_cols
    ).sort_values(ascending=False).head(20)
    fig, ax = plt.subplots(figsize=(7, 6))
    xgb_importances.plot(kind="barh", ax=ax)
    ax.invert_yaxis()
    ax.set_title(f"XGB — Top 20 Feature Importances (CV={CV_FOLDS})")
    ax.set_xlabel("Importance")
    plt.tight_layout()
    plt.savefig(fold_out_dir / "XGB_feature_importance.png", dpi=200)
    plt.show()

    # ──────────────────────────────────────────
    # Step 8: Save summary for this fold setting
    # ──────────────────────────────────────────
    all_metrics = rf_metrics_list + xgb_metrics_list
    summary_df  = pd.DataFrame(all_metrics)
    summary_df["CV_FOLDS"] = CV_FOLDS
    summary_df.to_csv(fold_out_dir / "model_comparison.csv", index=False)

    best_params_df = pd.DataFrame([
        {"model": "RF",  **rf_search.best_params_},
        {"model": "XGB", **xgb_search.best_params_}
    ])
    best_params_df.to_csv(fold_out_dir / "best_params.csv", index=False)

    print(f"\nModel comparison (CV_FOLDS={CV_FOLDS}):")
    print(summary_df.to_string(index=False))
    print(f"Saved to: {fold_out_dir}")

    all_fold_results[CV_FOLDS] = summary_df

# ── Final side-by-side comparison ──
print(f"\n{'='*60}")
print("  FINAL COMPARISON: CV=5 vs CV=10")
print(f"{'='*60}")
combined = pd.concat(all_fold_results.values()).sort_values(["CV_FOLDS", "model", "threshold"])
combined.to_csv(OUT_DIR / "model_comparison_all_folds.csv", index=False)
print(combined.to_string(index=False))
print(f"\nAll outputs saved under: {OUT_DIR}")
```
